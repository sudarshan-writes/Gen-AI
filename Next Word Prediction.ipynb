```python
data = """Machine learning is a field of artificial intelligence that focuses on building systems that learn from data. These systems improve their performance as they are exposed to more examples over time. A machine learning model identifies patterns in the data and uses those patterns to make predictions or decisions.

Supervised learning is one of the most common types of machine learning where the model is trained on labeled data. In this approach each input is associated with a correct output and the model learns to map inputs to outputs. Regression and classification are examples of supervised learning tasks.

Unsupervised learning deals with data that does not have labeled outputs. The goal is to find hidden patterns or structures in the data. Clustering and dimensionality reduction are common unsupervised learning techniques that help in understanding the data better.

Deep learning is a subset of machine learning that uses neural networks with multiple layers. These layers help the model learn complex representations of data. Deep learning has achieved remarkable success in areas such as computer vision and natural language processing.

Natural language processing enables machines to understand and generate human language. It is used in applications such as chatbots translation systems and sentiment analysis. Models learn from large text corpora to capture grammar context and meaning.

When training a neural network it is important to choose the right hyperparameters. The learning rate controls how quickly the model updates its weights. A very high learning rate may cause the model to diverge while a very low learning rate can slow down the training process.

Overfitting occurs when a model learns the training data too well including noise and outliers. This results in poor generalization to new data. Techniques such as regularization dropout and cross validation are used to prevent overfitting.

Underfitting happens when a model is too simple to capture the underlying structure of the data. In this case the model performs poorly on both training and testing data. Increasing model complexity or providing more relevant features can help address underfitting.

Data preprocessing is a crucial step in any machine learning pipeline. It in
volves cleaning the data handling missing values encoding categorical variables and scaling numerical features. Proper preprocessing ensures that the model learns effectively.

Feature engineering is the process of creating new features from existing data to improve model performance. Good features can significantly enhance the predictive power of a model. Domain knowledge often plays a key role in feature engineering.

In sequence models such as recurrent neural networks the order of data matters. These models process input sequences one step at a time while maintaining a hidden state. This hidden state captures information from previous time steps.

Long short term memory networks are designed to overcome the limitations of simple recurrent neural networks. They use gates to control the flow of information and maintain long term dependencies. This makes them suitable for tasks like text generation and speech recognition.

Gated recurrent units are a simplified version of long short term memory networks. They combine some of the gates to reduce computational complexity while still maintaining performance. GRU models are often faster to train.
  
In next word prediction the model learns to predict the next word given a sequence of previous words. For example given the phrase machine learning is the model may predict powerful or interesting depending on the training data.

Tokenization is the process of converting text into smaller units such as words or subwords. These tokens are then converted into numerical representations so that they can be processed by machine learning models.

Padding is used to ensure that all input sequences have the same length. This is important for batch processing in neural networks. Shorter sequences are padded with zeros to match the length of the longest sequence.

Embedding layers map words to dense vectors of fixed size. These vectors capture semantic relationships between words. Words with similar meanings tend to have similar vector representations.

The loss function measures how well the model predictions match the actual targets. During training the model updates its weights to minimize the loss. Optimization algorithms such as gradient descent are used for this purpose.

Backpropagation is the process of computing gradients of the loss function with respect to model parameters. These gradients are used to update the weights in the direction that reduces the loss.

Gradient descent is an optimization algorithm that iteratively updates model parameters. Variants such as stochastic gradient descent and Adam are commonly used in practice.

Evaluation metrics are used to measure the performance of a model."""
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([data])
```

```python
tokenizer.word_index
```

```python
input_seq = []
for paragraph in data.split('\n\n'):
    for sentence in paragraph.split('. '):
        tokenized_output = tokenizer.texts_to_sequences([sentence])[0]
       
    for i in range(1,len(tokenized_output)):
        input_seq.append(tokenized_output[:i+1])
```

```python
from keras.preprocessing.sequence import pad_sequences
```

```python
max_length = max([len(i) for i in input_seq])
```

```python
padded_data = pad_sequences(input_seq,maxlen =max_length, padding = 'pre')
```

```python
padded_data
```

```python
X = padded_data[:,:-1]
y = padded_data[:,-1]
```

```python
y
```

```python
from keras.utils import to_categorical
```

```python
len(tokenizer.word_index)
```

```python
X.shape
```

```python
y = to_categorical(y,num_classes=333) #Length of voc(x) + 1
```

## LSTM Architecture

```python
from keras.models import Sequential
from keras.layers import Input,Embedding,LSTM,Dense,Bidirectional,Dropout
```

```python
model = Sequential()
model.add(Input(shape=(22,)))
model.add(Embedding(input_dim=333,output_dim=100))
model.add(Bidirectional(LSTM(150)))
model.add(Dense(333,activation='softmax'))
```

```python
model.compile(optimizer='adam',
    loss = 'categorical_crossentropy',
    metrics=['accuracy',])
```

```python
model.fit(X,y,epochs=500)
```

```python
import numpy as np

# Function to predict next words given a seed sentence
def predict_next_word(seed_text, next_words=3):
    
    for _ in range(next_words):
        
        # Convert input text → sequence
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        
        # Pad sequence to match training input size
        token_list = pad_sequences([token_list], 
                                   maxlen=max_length-1, 
                                   padding='pre')
        
        # Predict probability distribution
        predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)[0]
        
        # Convert predicted index → word
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                seed_text += " " + word
                break
                
    return seed_text
```

```python
input_seq = []
for paragraph in data.split('\n\n'):
    for sentence in paragraph.split('. '):
        tokenized_output = tokenizer.texts_to_sequences([sentence])[0]
       
    for i in range(1,len(tokenized_output)):
        input_seq.append(tokenized_output[:i+1])
```

```python
max_length = max([len(i) for i in input_seq])
```

```python
import numpy as np

# Function to predict next words given a seed sentence
def predict_next_word(seed_text, next_words=3):
    
    for _ in range(next_words):
        
        # Convert input text → sequence
        token_list = tokenizer.texts_to_sequences([seed_text])[0]
        
        # Pad sequence to match training input size
        token_list = pad_sequences([token_list], 
                                   maxlen=max_length-1, 
                                   padding='pre')
        
        # Predict probability distribution
        predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)[0]
        
        # Convert predicted index → word
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                seed_text += " " + word
                break
                
    return seed_text
```
